## Colorbars

Using a suitable colorbar is an essential part of delivering information to its readers. The colorbar needs to be chosen such that it sets the information in a correct and understandable context, to transfer the important take-outs and to be clearly visible. 
When colorbars are used incorrectly they can falsify or even manipulate the data, e.g highlighting some data over others.

It is the goal to make plots easily and correctly understandable, also for the non-scientific reader. For example weather forecast and hazard maps are highly societal-relevant data, which need to tranfer the information clearly understandable to the wider public. Thus a well created plot is of main importance.

This tutorial introduces to use colorbars with the matplotlib library.

```python
import numpy as np 
import xarray as xr
import matplotlib.pyplot as plt # import matplotlib
import IPython.display as display
```

### Sequential Colormaps

Colors change in lightness and saturation, often using one hue. It should be used for representing information that is ordered (e.g. increasing/decreasing).

```python
display.Image(url="https://matplotlib.org/stable/_images/sphx_glr_colormap_reference_002.png")
```

Matplotlib provides colormaps that are perceived as sequential even though they are using multiple colors. 
Viridis is one of the most known colorbar, being the default for sequential information in matplotlib.

```python
display.Image(url="https://matplotlib.org/stable/_images/sphx_glr_colormap_reference_001.png")
```

More sequential colormaps within the matplotlib library can be found here: https://matplotlib.org/stable/gallery/color/colormap_reference.html

### Diverging colormaps

Diverging colormaps change in lighntess and saturation of two different colors, which meet in the middle at an unsaturated color.
They are used when the plotted information has a critical middle value (e.g. data deviating around zero)

```python
display.Image(url="https://matplotlib.org/stable/_images/sphx_glr_colormap_reference_004.png")
```

Note: To get reversed colorbars just add `_r` to the colorbar name. E.g. PiYG -> PiYG_r

### Cyclic colormaps

Usually two colors that change in lightness and meet at the beginning and end at an unsaturated color.

```python
display.Image(url="https://matplotlib.org/stable/_images/sphx_glr_colormap_reference_005.png")
```

### Qualitative Colormaps

Miscellaneous colors, which are used to represent information which have no order or relationship.

```python
display.Image(url="https://matplotlib.org/stable/_images/sphx_glr_colormap_reference_006.png")
```

### Miscellaneous Colormaps

Intentionally made for particular uses, e.g. ocean, gist_earth or terrain to differ between land and ocean. However they are not ideal for every use, so if selected they should be checked thoroughly.

```python
display.Image(url="https://matplotlib.org/stable/_images/sphx_glr_colormap_reference_007.png")
```

## Color vision deficiencies (CDV)

The most common form is the red-green color vision deficiency, where differentiating between red and green is not possible. Avoiding colormaps which inlcude both colors is already a pretty good step. 
Using a color palette with margentha (instead of red) and green, increases the contrast and help to differ between the colors.

In matplotlib `viridis` and `cividis` are colorblind friendly, whereas `jet` is know to be problematic (and still widley used). 

Here you see an example for the colorbar "Temps" and how it is seen for people with different CDV. It is not distinguishable for persons with euteranopia or protanopia.

```python
display.Image(url="https://jakubnowosad.com/colorblindcheck/articles/check_rcartocolor_files/figure-html/unnamed-chunk-27-1.png", 
              width= 600,
              height= 500)

# Colorbar "Temps" is not a good choice for people with CDV
```

## Plotting Colorbars

### Sequential and diverging colorbars

Now we look at a practial example:

```python
file1 = "C:/your/path/25_ERA5_reanalysis_sfc_2025-11-16_2025-11-27.nc"
ds1 = xr.open_dataset(file1)
ds1 
```

When using the .plot() command, it automatically generates a colorbar. This is a smart feature especially as it differs between sequential and divergent values. 
Still we want to have more freedom and adapting our own colorbar when creating a plot by using the matplotlib library.


```python
# Plotting the u10-wind is automatically plotted with a diverging colorbar
u10_plot = ds1["u10"].isel(time = 0)
u10_plot.plot();
```

Let's adapat the colorbar by ourselfes:

```python
import cartopy.crs as ccrs # for geographic plotting
import cartopy.feature as cfeature
from cmcrameri import cm    # library which provides more colorbars
```

The wind variable comes with positive (eastward wind) and negative values (westward wind). Thus a diverging colorbar is needed.
If we nevertheless select a sequential colorbar (e.g. viridis) we lose important information and don't depict the information properly:

```python
fig = plt.figure(figsize=(9, 5))
ax = plt.axes(projection=ccrs.PlateCarree())

u10_plot.plot.contourf(ax = ax, cmap = "viridis") 
ax.add_feature(cfeature.COASTLINE, linestyle='-')
ax.add_feature(cfeature.BORDERS, linestyle='-', linewidth=.5)
ax.set_title("Sequential colorbar", fontsize = 15);
```

Instead we can chose a divergent colorbar and we get a plot which clearly shows the wind direction.

The chosen color for this variable is recommended by the IPCC. Each variable has its own colorcoding. 
More information can be found here: https://www.ipcc.ch/site/assets/uploads/2022/09/IPCC_AR6_WGI_VisualStyleGuide_2022.pdf (p. 11)

```python
fig = plt.figure(figsize=(9, 5))
ax = plt.axes(projection=ccrs.PlateCarree())

u10_plot.plot.contourf(ax = ax, cmap = cm.cork) # cork is the recommended colorbar by the IPCC for diverging wind
ax.add_feature(cfeature.COASTLINE, linestyle='-')
ax.add_feature(cfeature.BORDERS, linestyle='-', linewidth=.5)
ax.set_title("Diverging colorbar", fontsize = 15);
```

### Number of levels

Select a good amount of levels (not too few, not too many). It depends on the context but selecting 7-11 levels is ususally a good choice.


```python
fig = plt.figure(figsize=(9, 5))
ax = plt.axes(projection=ccrs.PlateCarree())
lv = 7

u10_plot.plot.contourf(ax = ax, cmap = cm.cork, levels = lv)
ax.add_feature(cfeature.COASTLINE, linestyle='-')
ax.add_feature(cfeature.BORDERS, linestyle='-', linewidth=.5)
ax.set_title(f"{lv} levels", fontsize = 15);
```

Changing the level to 9 changes can adjust the range of the colorbar on the right site but also the patterns within the graph (e.g. compare the darkblue patch east of Vietnam).
The information put into both plots are still the same, but they are depicted differently depending on the levels you allow.

```python
fig = plt.figure(figsize=(9, 5))
ax = plt.axes(projection=ccrs.PlateCarree())
lv = 9

u10_plot.plot.contourf(ax = ax, cmap = cm.cork, levels = lv) # cork is the recommended colorbar by the IPCC
ax.add_feature(cfeature.COASTLINE, linestyle='-')
ax.add_feature(cfeature.BORDERS, linestyle='-', linewidth=.5)
ax.set_title(f"{lv} levels", fontsize = 15);
```

```python
fig = plt.figure(figsize=(9, 5))
ax = plt.axes(projection=ccrs.PlateCarree())
lv = 11

u10_plot.plot.contourf(ax = ax, cmap = cm.cork, levels = lv) # cork is the recommended colorbar by the IPCC
ax.add_feature(cfeature.COASTLINE, linestyle='-')
ax.add_feature(cfeature.BORDERS, linestyle='-', linewidth=.5)
ax.set_title(f"{lv} levels", fontsize = 15);
```

Lets make the it more extreme and put 31 levels. You see that the contours become finer. Allowing more levels looks nice in the graph, but with a look at the colorbar on the right you see it is also more difficult to say which color depicts which value. You need to find the compromise between level resolution and readability of the plot.


```python
fig = plt.figure(figsize=(9, 5))
ax = plt.axes(projection=ccrs.PlateCarree())
lv = 31

u10_plot.plot.contourf(ax = ax, cmap = cm.cork, levels = lv) # cork is the recommended colorbar by the IPCC
ax.add_feature(cfeature.COASTLINE, linestyle='-')
ax.add_feature(cfeature.BORDERS, linestyle='-', linewidth=.5)
ax.set_title(f"{lv} levels", fontsize = 15);
```

### Projections

The cartopy library adds the feature of plotting our geographical data on different projections. 
Chosing the a fitting projection helps to understand the plot better. 

- ``PlateCarree()``: is the classic projection, which is great for looking at a selected area (recommended).
- ``NorthPolarStereo()``: In some cases you have information for the whole Northern Hemisphere, then this projection adds a lot of value.
- ``SouthPolarStereo()``: Accordingly for the Southern Hemisphere.
- `Orthographic()`: Another nice projection.        

An overview of available projections can be found here: https://scitools.org.uk/cartopy/docs/v0.15/crs/projections.html

```python
# Load data which contain information of the whole Northern Hemisphere
file2 = "C:/your/path/CRA_course_support/Climatology_Mean_z500_DJF_1990-2019_1x1.nc"
ds2 = xr.open_dataset(file2)
z_plot = ds2["geopotential"]
```

```python
fig = plt.figure(figsize=(9, 5))
ax = plt.axes(projection=ccrs.PlateCarree())

z_plot.plot.contourf(ax = ax) 
ax.add_feature(cfeature.COASTLINE, linestyle='-')
ax.add_feature(cfeature.BORDERS, linestyle='-', linewidth=.5)
ax.set_title("PlateCarree projection", fontsize = 15);
```

```python
fig = plt.figure(figsize=(9, 5))
ax = plt.axes(projection=ccrs.NorthPolarStereo()) # Change the projection in the axis configurations

z_plot.plot.contourf(ax = ax, transform=ccrs.PlateCarree()) # add(!) transform = PlateCarree projection. Data are naive to the source and destination coordinate systems which is usually PlateCarree(). Otherwise the plot won't work.
ax.add_feature(cfeature.COASTLINE, linestyle='-')
ax.add_feature(cfeature.BORDERS, linestyle='-', linewidth=.5)
ax.set_title("NorthPolarStereo projection", fontsize = 15);
```

```python
fig = plt.figure(figsize=(9, 5))
ax = plt.axes(projection=ccrs.Orthographic()) # Change the projection in the axis configurations

z_plot.plot.contourf(ax = ax, transform=ccrs.PlateCarree()) # add(!) transform = PlateCarree projection. Data are naive to the source and destination coordinate systems which is usually PlateCarree(). Otherwise the plot won't work.
ax.add_feature(cfeature.COASTLINE, linestyle='-')
ax.add_feature(cfeature.BORDERS, linestyle='-', linewidth=.5)
ax.set_title("Orthographic projection", fontsize = 15);
```

**Advanced**: You might have noticed the white gap in the data when plotting. This is a common "bug" because the data cover 0-360°, but not the connection from 360° to 0°. So it is not plotted as a closed area. 
For this cartopy provides a feature called `add_cyclic_point`. 

```python
from cartopy.util import add_cyclic_point
```

```python
fig = plt.figure(figsize=(9, 5))
ax = plt.axes(projection=ccrs.Orthographic())  

# Add cyclic points to your data
data_cyc_closed, data_cyc_lon = add_cyclic_point(z_plot, coord=z_plot.longitude)    # You add your data and the data's longitude. 
                                                                                    # As output you get your closed data (data_cyc_closed) as well as your extended longitude which bridges the gap from 360° to 0° (data_cyc_lon)

img = ax.contourf(data_cyc_lon, z_plot.latitude, data_cyc_closed, transform=ccrs.PlateCarree())     # Note that contourf is adressed here via the ax-feature here. It does the same as before. 
                                                                                                    # Is is just more convenient to address the input more specifially. Data, lon, lat = data_cyc_lon, z_plot.latitude, data_cyc_closed
ax.add_feature(cfeature.COASTLINE, linestyle='-')
ax.add_feature(cfeature.BORDERS, linestyle='-', linewidth=.5)
ax.set_title("Orthographic projection", fontsize = 15)

cbar = plt.colorbar(img, ax=ax, orientation='vertical', shrink=0.8, fraction=0.18, pad=0.05, aspect=20); # due to applying contourf via the ax feature, the colorbar has to be added extra
```
